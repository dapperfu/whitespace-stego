# Whitespace Steganography Example Notebook

This notebook demonstrates how to use the `whitespace_stego` library to encode and decode messages using zero-width Unicode characters. This technique allows you to hide secret messages within seemingly normal text without changing its visual appearance.

## What is Whitespace Steganography?

Whitespace steganography is a technique that hides information in text by using invisible Unicode characters. The hidden message is embedded using zero-width characters that are not visible to the human eye but can be detected and decoded by software.

### Key Features:
- **Invisible**: The hidden message doesn't change the visual appearance of the text
- **Secure**: Optional password encryption using Fernet (AES-128)
- **Flexible**: Can embed messages in any text or use standalone encoded strings
- **Robust**: Uses base64 encoding and proper error handling

## Installation and Setup

First, let's make sure we have the library installed and import the necessary modules.

```python
# Import the whitespace steganography library
from whitespace_stego import core

# For demonstration purposes, we'll also import some utility functions
import base64
import hashlib
from typing import Optional

print("✅ Whitespace steganography library imported successfully!")
print(
    f"Available functions: {[func for func in dir(core) if not func.startswith('_')]}"
)
```

## Basic Encoding and Decoding

Let's start with the simplest example: encoding a message without a carrier text.

```python
# Example 1: Basic encoding without carrier text
message = "Hello, this is a secret message!"

# Encode the message
encoded = core.encode(message)

print("Original message:")
print(f"'{message}'")
print(f"Length: {len(message)} characters")
print()

print("Encoded message (zero-width characters):")
print(f"'{encoded}'")
print(f"Length: {len(encoded)} characters")
print()

# Decode the message
decoded = core.decode(encoded)

print("Decoded message:")
print(f"'{decoded}'")
print()

# Verify the roundtrip worked
print(f"✅ Roundtrip successful: {message == decoded}")
```

## Encoding with Carrier Text

Now let's embed a message within a carrier text. The message will be hidden after the first character of the carrier.

```python
# Example 2: Encoding with carrier text
secret_message = "Meet me at the coffee shop at 3 PM"
carrier_text = "This is just a normal email about the quarterly report."

# Encode the secret message into the carrier
encoded_carrier = core.encode(secret_message, carrier_text)

print("Original carrier text:")
print(f"'{carrier_text}'")
print()

print("Secret message to hide:")
print(f"'{secret_message}'")
print()

print("Encoded carrier (looks identical to original):")
print(f"'{encoded_carrier}'")
print()

# Verify they look the same
print(f"Visual comparison - Same length: {len(carrier_text) == len(encoded_carrier)}")
print(f"Visual comparison - Same start: {carrier_text[:20] == encoded_carrier[:20]}")
print()

# Decode the hidden message
decoded_secret = core.decode(encoded_carrier)

print("Decoded secret message:")
print(f"'{decoded_secret}'")
print()

print(f"✅ Secret message recovered: {secret_message == decoded_secret}")
```

## Password Protection

Let's add encryption to our messages using a password. This provides an additional layer of security.

```python
# Example 3: Password-protected encoding
sensitive_message = "Bank account: 1234-5678-9012-3456, PIN: 1234"
password = "my_secret_password_123"
carrier = "Please review the attached documents for the upcoming meeting."

# Encode with password protection
encrypted_carrier = core.encode(sensitive_message, carrier, password)

print("Sensitive message:")
print(f"'{sensitive_message}'")
print()

print("Password-protected carrier:")
print(f"'{encrypted_carrier}'")
print()

# Try to decode without password (should fail)
try:
    wrong_decode = core.decode(encrypted_carrier)
    print("❌ This should have failed!")
except ValueError as e:
    print(f"✅ Correctly failed without password: {e}")
print()

# Decode with correct password
decoded_sensitive = core.decode(encrypted_carrier, password)

print("Decoded with correct password:")
print(f"'{decoded_sensitive}'")
print()

print(f"✅ Password protection works: {sensitive_message == decoded_sensitive}")
```

## Error Handling and Validation

Let's see how the library handles various error conditions.

```python
# Example 4: Error handling demonstrations

print("=== Error Handling Examples ===\n")

# Test 1: Decoding text without any hidden message
normal_text = "This is just normal text without any hidden content."
try:
    result = core.decode(normal_text)
    print("❌ Should have failed!")
except ValueError as e:
    print(f"✅ Correctly failed to decode normal text: {e}")
print()

# Test 2: Wrong password
secret = "My secret message"
carrier = "Normal carrier text"
correct_password = "correct123"
wrong_password = "wrong456"

encoded = core.encode(secret, carrier, correct_password)

try:
    result = core.decode(encoded, wrong_password)
    print("❌ Should have failed with wrong password!")
except ValueError as e:
    print(f"✅ Correctly failed with wrong password: {e}")
print()

# Test 3: Empty message
empty_message = ""
carrier = "Test carrier"

try:
    encoded = core.encode(empty_message, carrier)
    decoded = core.decode(encoded)
    print(f"✅ Empty message handled correctly: '{decoded}'")
except Exception as e:
    print(f"❌ Failed to handle empty message: {e}")
```

## Real-World Applications

Let's create some practical examples that could be used in real scenarios.

```python
# Example 6: Digital signature verification
import json
from datetime import datetime


def create_signed_message(content: str, author: str, private_key: str) -> str:
    """Create a message with a digital signature embedded"""
    timestamp = datetime.now().isoformat()

    # Create signature data
    signature_data = {
        "content_hash": hashlib.sha256(content.encode()).hexdigest(),
        "author": author,
        "timestamp": timestamp,
        "version": "1.0",
    }

    # Create signature
    signature_string = json.dumps(signature_data, sort_keys=True)
    signature_hash = hashlib.sha256(
        (signature_string + private_key).encode()
    ).hexdigest()

    # Embed signature in content
    full_signature = json.dumps({"signature": signature_hash, "data": signature_data})

    return core.encode(full_signature, content, private_key)


def verify_signed_message(
    encoded_content: str, private_key: str
) -> tuple[bool, str, dict]:
    """Verify a signed message and return verification result"""
    try:
        # Extract signature
        signature_json = core.decode(encoded_content, private_key)
        signature_data = json.loads(signature_json)

        # Extract original content
        original_content, _ = core.extract_encoded(encoded_content)

        # Verify signature
        data_part = json.dumps(signature_data["data"], sort_keys=True)
        expected_hash = hashlib.sha256((data_part + private_key).encode()).hexdigest()

        is_valid = signature_data["signature"] == expected_hash

        return is_valid, original_content, signature_data["data"]

    except Exception as e:
        return False, str(e), {}


# Test digital signature
document = "This is an important business document that needs verification."
author = "john.doe@company.com"
private_key = "company_secret_key_2024"

print("=== Digital Signature Example ===\n")

# Create signed document
signed_doc = create_signed_message(document, author, private_key)
print(f"Original document: '{document}'")
print(f"Signed document length: {len(signed_doc)}")
print()

# Verify signature
is_valid, extracted_content, signature_info = verify_signed_message(
    signed_doc, private_key
)
print(f"Signature valid: {is_valid}")
print(f"Extracted content: '{extracted_content}'")
print(f"Signature info: {signature_info}")
print()

# Test with wrong key
is_valid_wrong, _, _ = verify_signed_message(signed_doc, "wrong_key")
print(f"Verification with wrong key: {is_valid_wrong}")
```
