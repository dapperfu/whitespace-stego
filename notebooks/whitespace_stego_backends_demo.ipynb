# Whitespace Steganography: Python vs Rust Backends Demo

This notebook demonstrates how both the Python and Rust backends of the `whitespace_stego` library can encode and decode identical messages. This showcases the interoperability between the two implementations.

## What is Whitespace Steganography?

Whitespace steganography is a technique that hides information in text using invisible Unicode characters. The hidden message is embedded using zero-width characters that are not visible to the human eye but can be detected and decoded by software.

### Key Features:
- **Invisible**: The hidden message doesn't change the visual appearance of the text
- **Secure**: Optional password encryption using Fernet (AES-128)
- **Flexible**: Can embed messages in any text or use standalone encoded strings
- **Robust**: Uses base64 encoding and proper error handling
- **Multi-backend**: Both Python and Rust implementations available

## Installation and Setup

First, let's make sure we have both backends available and import the necessary modules.

```python
# Import both backends
from whitespace_stego import core as python_backend
from whitespace_stego_backend import encode as rust_encode, decode as rust_decode

# For demonstration purposes, we'll also import some utility functions
import base64
import hashlib
from typing import Optional, Tuple
import time

print("✅ Python backend imported successfully!")
print(
    f"Python functions: {[func for func in dir(python_backend) if not func.startswith('_')]}"
)
print()
print("✅ Rust backend imported successfully!")
print(f"Rust functions: encode, decode")
print()
print("Both backends are ready for comparison!")
```

## Backend Comparison: Basic Encoding and Decoding

Let's start by comparing how both backends handle basic encoding and decoding without carrier text.

```python
# Test message
message = "Hello, this is a secret message from both backends!"

print("Original message:")
print(f"'{message}'")
print(f"Length: {len(message)} characters")
print()

# Encode with Python backend
python_encoded = python_backend.encode(message)
print("Python backend encoded:")
print(f"'{python_encoded}'")
print(f"Length: {len(python_encoded)} characters")
print()

# Encode with Rust backend
rust_encoded = rust_encode(message, "", None)
print("Rust backend encoded:")
print(f"'{rust_encoded}'")
print(f"Length: {len(rust_encoded)} characters")
print()

# Compare the encoded outputs
print("🔍 Backend Comparison:")
print(f"Encoded outputs are identical: {python_encoded == rust_encoded}")
print(f"Python encoded length: {len(python_encoded)}")
print(f"Rust encoded length: {len(rust_encoded)}")
print()

# Decode with both backends
python_decoded = python_backend.decode(python_encoded)
rust_decoded = rust_decode(rust_encoded, None)

print("Decoded messages:")
print(f"Python decoded: '{python_decoded}'")
print(f"Rust decoded: '{rust_decoded}'")
print()

# Verify roundtrip
print("✅ Roundtrip verification:")
print(f"Python roundtrip successful: {message == python_decoded}")
print(f"Rust roundtrip successful: {message == rust_decoded}")
print(f"Cross-backend compatibility: {python_decoded == rust_decoded}")
```

## Cross-Backend Decoding Test

Now let's test if each backend can decode messages encoded by the other backend.

```python
# Test cross-backend compatibility
test_message = "Cross-backend compatibility test message!"

print("🔄 Cross-Backend Compatibility Test")
print("=" * 50)
print()

# Python encodes, Rust decodes
print("1. Python encodes → Rust decodes:")
py_encoded = python_backend.encode(test_message)
rust_decoded_from_py = rust_decode(py_encoded, None)
print(f"   Success: {test_message == rust_decoded_from_py}")
print(f"   Decoded: '{rust_decoded_from_py}'")
print()

# Rust encodes, Python decodes
print("2. Rust encodes → Python decodes:")
rust_encoded = rust_encode(test_message, "", None)
py_decoded_from_rust = python_backend.decode(rust_encoded)
print(f"   Success: {test_message == py_decoded_from_rust}")
print(f"   Decoded: '{py_decoded_from_rust}'")
print()

print("✅ Cross-backend compatibility verified!")
```

## Encoding with Carrier Text

Let's test both backends with carrier text to see if they produce identical results.

```python
# Test with carrier text
secret_message = "Meet me at the coffee shop at 3 PM"
carrier_text = "This is just a normal email about the quarterly report."

print("📧 Carrier Text Encoding Test")
print("=" * 40)
print()

print("Original carrier text:")
print(f"'{carrier_text}'")
print()

print("Secret message to hide:")
print(f"'{secret_message}'")
print()

# Encode with both backends
python_carrier = python_backend.encode(secret_message, carrier_text)
rust_carrier = rust_encode(secret_message, carrier_text, None)

print("Encoded carriers (first 50 chars):")
print(f"Python: '{python_carrier[:50]}...'")
print(f"Rust:   '{rust_carrier[:50]}...'")
print()

# Compare outputs
print("🔍 Carrier Comparison:")
print(f"Encoded carriers are identical: {python_carrier == rust_carrier}")
print(f"Python carrier length: {len(python_carrier)}")
print(f"Rust carrier length: {len(rust_carrier)}")
print()

# Decode with both backends
python_decoded_secret = python_backend.decode(python_carrier)
rust_decoded_secret = rust_decode(rust_carrier, None)

print("Decoded secret messages:")
print(f"Python decoded: '{python_decoded_secret}'")
print(f"Rust decoded: '{rust_decoded_secret}'")
print()

print("✅ Carrier text roundtrip verification:")
print(f"Python roundtrip: {secret_message == python_decoded_secret}")
print(f"Rust roundtrip: {secret_message == rust_decoded_secret}")
print(f"Cross-backend: {python_decoded_secret == rust_decoded_secret}")
```

## Password Protection Comparison

Let's test password-protected encoding and decoding with both backends.

```python
# Test password protection
sensitive_message = "Bank account: 1234-5678-9012-3456, PIN: 1234"
password = "my_secret_password_123"
carrier = "Please review the attached documents for the upcoming meeting."

print("🔐 Password Protection Test")
print("=" * 35)
print()

print("Sensitive message:")
print(f"'{sensitive_message}'")
print()

print("Password: 'my_secret_password_123'")
print()

# Encode with password using both backends
python_encrypted = python_backend.encode(sensitive_message, carrier, password)
rust_encrypted = rust_encode(sensitive_message, carrier, password)

print("Encrypted carriers (first 50 chars):")
print(f"Python: '{python_encrypted[:50]}...'")
print(f"Rust:   '{rust_encrypted[:50]}...'")
print()

# Compare encrypted outputs
print("🔍 Encrypted Output Comparison:")
print(f"Encrypted carriers are identical: {python_encrypted == rust_encrypted}")
print()

# Try to decode without password (should fail)
print("❌ Testing decode without password:")
try:
    python_wrong = python_backend.decode(python_encrypted)
    print("   Python: This should have failed!")
except ValueError as e:
    print(f"   Python: Correctly failed - {e}")

try:
    rust_wrong = rust_decode(rust_encrypted, None)
    print("   Rust: This should have failed!")
except ValueError as e:
    print(f"   Rust: Correctly failed - {e}")
print()

# Decode with correct password
python_decrypted = python_backend.decode(python_encrypted, password)
rust_decrypted = rust_decode(rust_encrypted, password)

print("✅ Decrypted messages:")
print(f"Python decrypted: '{python_decrypted}'")
print(f"Rust decrypted: '{rust_decrypted}'")
print()

print("🔐 Password protection verification:")
print(f"Python roundtrip: {sensitive_message == python_decrypted}")
print(f"Rust roundtrip: {sensitive_message == rust_decrypted}")
print(f"Cross-backend: {python_decrypted == rust_decrypted}")
```

## Performance Comparison

Let's compare the performance of both backends with larger messages.

## C Backend Integration

Now let's demonstrate the C backend functionality alongside Python and Rust backends. The C implementation provides a command-line interface for encoding and decoding messages.

```python
# C Backend demonstration
import subprocess
import tempfile
import os

print("🔧 C Backend Integration Test")
print("=" * 40)
print()

# Check if C backend is available
c_backend_available = os.path.exists('./whitespace-stego-c')
if not c_backend_available:
    print("⚠️  C backend binary not found. C examples will be skipped.")
    print("Make sure to run this notebook from the project root directory.")
else:
    print("✅ C backend binary found!")
    print()

    # Test message and carrier
    c_test_message = "Hello from C backend!"
    c_test_carrier = "This is a carrier text for C backend testing."

    print(f"Test message: '{c_test_message}'")
    print(f"Test carrier: '{c_test_carrier}'")
    print()

    # Create temporary files for C backend
    with tempfile.NamedTemporaryFile(mode='w', suffix='.txt', delete=False) as msg_file:
        msg_file.write(c_test_message)
        msg_file_path = msg_file.name

    with tempfile.NamedTemporaryFile(mode='w', suffix='.txt', delete=False) as carrier_file:
        carrier_file.write(c_test_carrier)
        carrier_file_path = carrier_file.name

    with tempfile.NamedTemporaryFile(mode='w', suffix='.txt', delete=False) as output_file:
        output_file_path = output_file.name

    print("📁 Temporary files created:")
    print(f"   Message file: {msg_file_path}")
    print(f"   Carrier file: {carrier_file_path}")
    print(f"   Output file: {output_file_path}")
    print()

    # Test C backend encoding
    print("1. C Backend Encoding:")
    try:
        # Encode using C backend
        encode_cmd = [
            './whitespace-stego-c', 'encode',
            '--message', msg_file_path,
            '--carrier', carrier_file_path,
            '--output', output_file_path
        ]
        
        result = subprocess.run(encode_cmd, capture_output=True, text=True, timeout=10)
        
        if result.returncode == 0:
            print("   ✅ C encoding successful!")
            
            # Read the encoded output
            with open(output_file_path, 'r') as f:
                c_encoded = f.read()
            
            print(f"   Encoded length: {len(c_encoded)} characters")
            print(f"   Contains zero-width chars: {'\\u200b' in c_encoded or '\\u200d' in c_encoded}")
            
        else:
            print(f"   ❌ C encoding failed: {result.stderr}")
            
    except Exception as e:
        print(f"   ❌ C encoding error: {e}")
    print()

    # Test C backend decoding
    print("2. C Backend Decoding:")
    try:
        # Create a temporary file for decoded output
        with tempfile.NamedTemporaryFile(mode='w', suffix='.txt', delete=False) as decoded_file:
            decoded_file_path = decoded_file.name
        
        # Decode using C backend
        decode_cmd = [
            './whitespace-stego-c', 'decode',
            '--input', output_file_path,
            '--output', decoded_file_path
        ]
        
        result = subprocess.run(decode_cmd, capture_output=True, text=True, timeout=10)
        
        if result.returncode == 0:
            print("   ✅ C decoding successful!")
            
            # Read the decoded output
            with open(decoded_file_path, 'r') as f:
                c_decoded = f.read()
            
            print(f"   Decoded message: '{c_decoded}'")
            print(f"   Roundtrip successful: {c_test_message == c_decoded}")
            
        else:
            print(f"   ❌ C decoding failed: {result.stderr}")
            
    except Exception as e:
        print(f"   ❌ C decoding error: {e}")
    print()

    # Clean up temporary files
    try:
        os.unlink(msg_file_path)
        os.unlink(carrier_file_path)
        os.unlink(output_file_path)
        os.unlink(decoded_file_path)
        print("🧹 Temporary files cleaned up")
    except:
        pass
    print()
```
